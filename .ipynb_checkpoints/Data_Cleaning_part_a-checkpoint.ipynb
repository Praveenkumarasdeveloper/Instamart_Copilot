## 3. Clean customers (3 marks). 

1. (a) Standardise city by removing leading/trailing spaces and applying title case so exactly six cities remain. 
2. (b) Map every gender spelling to male, female, or unknown (blank → unknown). 
3. (c) Treat ages outside 10–100 as missing, then fill all missing ages with the median age of the customer's city, computed with groupby and transform.
4. (d) Convert signup_date to a datetime.

**Reference after cleaning:** 

    cities Pune 227, Delhi 210, Bengaluru 197, Chennai 191, Hyderabad 191, Mumbai 184; genders female 610, male 560, unknown 30; 0 missing ages and every age between 18 and 60.

# (a) Clean_City_column

```python
customer_detail_copy = Customer_details.copy()
```

```python
customer_detail_copy["city"].unique()
```

```python
customer_detail_copy["city"].unique()
```

```python
customer_detail_copy["city"]= (customer_detail_copy["city"].str.strip()
                                                           .str.title())
print(customer_detail_copy["city"])
print(customer_detail_copy["city"].unique())
```

# (b) Clean_Gender_column

```python
customer_detail_copy["gender"].unique()
```

```python
customer_detail_copy["gender"] = customer_detail_copy["gender"].astype(str).str.strip().str.title()
```

```python
customer_detail_copy["gender"]= customer_detail_copy["gender"].replace({"F":"Female",
                                       "M":"Male"}).fillna("unknown")
                              
```

```python
customer_detail_copy["gender"].unique()
```

```python
print(customer_detail_copy["gender"].value_counts())
```

# (c) Clean_Age_Column

```python
customer_detail_copy["age"].median()
```

```python
customer_detail_copy["age"] = pd.to_numeric(customer_detail_copy["age"],errors = "coerce")

customer_detail_copy.loc[~customer_detail_copy["age"].between(10,100),"age"] = np.nan

customer_detail_copy["age"] = customer_detail_copy["age"].fillna(
                                                                customer_detail_copy.groupby("city")["age"]
                                                                .transform("median"))

customer_detail_copy["age"]= customer_detail_copy["age"].astype(int)
```

# (d) Clean_the_date & time column

```python
customer_detail_copy["signup_date"] = pd.to_datetime(customer_detail_copy["signup_date"])
```

```python
customer_detail_copy
```

```python
print("Number_of_entries_City_wise : \n",customer_detail_copy["city"].value_counts(),"\n")

print("Number_of_entries_Gender_wise :\n",customer_detail_copy["gender"].value_counts(),"\n")

print("Number_of_Null_entries_whole_customer_table :\n",customer_detail_copy.isnull().sum(),"\n")
```

# 4. Clean orders (5 marks), in this order, documenting each rule in the README:

1. Convert order_ts to a datetime.

2. Remove exact duplicate rows.

3. Several orders still appear more than once with different statuses (an earlier pending record and a later final record). For each order_id keep only the record with the latest order_ts.

4. Convert order_amount to numeric, turning unparseable entries into missing values, then fill each missing amount from the product price: order_amount = round(price_inr × quantity × (1 − discount_pct / 100), 2).

5. Set negative delivery_minutes to missing. For orders with status completed, fill every missing delivery_minutes with the median delivery time of that order's product category among completed orders (use groupby and transform). For orders with status cancelled, leave delivery_minutes missing — a cancelled order was never delivered, so imputing a delivery time for it would invent data.

# 4.4 _ Clean_The_order_Amount_Column

**Convert order_amount to numeric, turning unparseable entries into missing values, then fill each missing amount from the product price:**

order_amount = round(price_inr × quantity × (1 − discount_pct / 100), 2).



```python
order_details_copy
```

```python
order_details_copy["order_amount"]= pd.to_numeric(
                                                    order_details_copy["order_amount"],
                                                    errors = "coerce")
```

## Using_the_Map

**Reason** -- it is join the product_detail and the order_details_copy table.

```python
product_details
```

```python
price_mapping = product_details.set_index("product_id")["price_inr"]
```

```python
order_details_copy["price_inr"] = (order_details_copy["product_id"].map(price_mapping))
```

```python
display(order_details_copy[["product_id","quantity","quantity","price_inr"]].head())
```

```python
order_details_copy.head()
```

```python
order_details_copy.dtypes
```

```python
pd.to_numeric(order_details_copy["discount_pct"],errors = "coerce")
```

```python
missing_amount = order_details_copy["order_amount"].isna()
```

```python
order_details_copy.loc[missing_amount,"order_amount"] = (
                                                        order_details_copy.loc[missing_amount,"price_inr"] 
                                                        * order_details_copy.loc[missing_amount,"quantity"]
                                                        *(1 - order_details_copy.loc[missing_amount,"discount_pct"] / 100 )
                                                        ).round(2)

print("Number_of_The_Missing_Amount_After_Cleaning : ", order_details_copy["order_amount"].isna().sum())
```

```python
order_details_copy["order_amount"].isna().sum()
```

# 4.5 _ Clean_Negative_Minute 

```python
order_details_copy.head()
```

```python
order_details_copy.loc[
    order_details_copy["delivery_minutes"] < 0,
    "delivery_minutes"
] = np.nan
```

```python
category_map = product_details.set_index("product_id")["category"]

order_details_copy["category"] = (order_details_copy["product_id"].map(category_map))

display(order_details_copy[["product_id","category","delivery_minutes","status"]].head())
```

```python
completed_mask = order_details_copy["status"].eq("completed")

order_details_copy.loc[completed_mask, "delivery_minutes"] = (
                                                                order_details_copy.loc[completed_mask, "delivery_minutes"].fillna(
                                                                order_details_copy.loc[completed_mask]
                                                                .groupby("category")["delivery_minutes"]
                                                                .transform("median")
                                                                    )
                                                                )
```

```python
completed = order_details_copy[order_details_copy["status"] == "completed"]
```

```python
cancelled = order_details_copy[order_details_copy["status"] == "cancelled"]
```
